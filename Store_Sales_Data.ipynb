Step 9: Investigate Duplicate Product IDs

```python
product_id_duplicates = dim_product[
    dim_product["ProductID"].duplicated(keep=False)
]

product_id_duplicates.sort_values("ProductID")
```

Step 10: Validate ProductKey Uniqueness

The ProductKey column was checked to ensure that each product has a unique identifier.

The result shows 0 duplicated values, confirming that ProductKey can be used as the primary key for the Dim_Product table.

Although duplicate values were found in ProductID, the ProductKey column remains unique. Therefore, the duplicate ProductID values are treated as source data inconsistencies and no records were removed.

```python
dim_product["ProductKey"].duplicated().sum()
```

Step 11: Analyze Product Categories

```python
dim_product["SubCategoryID"].value_counts()
```

Step 12: Check Product Price Values

```python
dim_product[["Unit_Cost", "List_Price"]].describe()
```

Step 13: Detect Price Outliers

```python
Q1 = dim_product["List_Price"].quantile(0.25)
Q3 = dim_product["List_Price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = dim_product[
    (dim_product["List_Price"] < lower_bound) |
    (dim_product["List_Price"] > upper_bound)
]

outliers
```
